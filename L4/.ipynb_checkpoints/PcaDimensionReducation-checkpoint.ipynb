```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
X, y = make_blobs(n_samples=500, centers=3, n_features=5,cluster_std=0.60,random_state=42)
```

```python
X
```

```python
scaler = StandardScaler()
X_scaler = scaler.fit_transform(X)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaler)
```

```python
df_pca = pd.DataFrame(X_pca,columns=["PC1","PC2"])
df_pca["label"] = y
```

```python
sns.scatterplot(data=df_pca,x="PC1",y="PC2",hue="label",palette="Set2")
```

### from above graph ,we can see that we have reduced the dimention from 5--> 2
